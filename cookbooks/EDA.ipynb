```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
en_train  = pd.read_csv('/kaggle/input/hope-speech-detection-across-multiple-languages/Training_Data_PolyHope-M_2025/Training Data/en_train.csv')
en_train
```

```python
en_dev  = pd.read_csv('/kaggle/input/hope-speech-detection-across-multiple-languages/Training_Data_PolyHope-M_2025/Training Data/en_dev.csv')
en_dev
```

```python
def plot_distribution(df, column_name, title):
    plt.figure(figsize=(14, 6))
    
    # Bar plot
    plt.subplot(1, 2, 1)
    sns.countplot(x=column_name, data=df, palette='viridis')
    plt.title(f'{title} - Bar Plot')
    plt.xticks(rotation=45)
    
    # Pie chart
    plt.subplot(1, 2, 2)
    df[column_name].value_counts().plot.pie(autopct='%1.1f%%', colors=sns.color_palette('viridis'))
    plt.title(f'{title} - Pie Chart')
    plt.ylabel('')
    
    plt.tight_layout()
    plt.show()
```

```python
# Plot for binary column
plot_distribution(en_train, 'binary', 'Binary Column Label Distribution')
```

```python
# Plot for multiclass column
plot_distribution(en_train, 'multiclass', 'Multiclass Column Label Distribution')
```

```python
# Plot for binary column
plot_distribution(en_dev, 'binary', 'Binary Column Label Distribution')
# Plot for multiclass column
plot_distribution(en_dev, 'multiclass', 'Multiclass Column Label Distribution')
```
